```python
library(emmeans)
library("lmtest")
library(car)
```

```python
df <- read.csv("output/discovery/discovery_cohort_metadata_SBSABCD_20261006.csv")
```

```python
df$PE_binary <- factor(df$PE_binary)
```

# 1B, 1C - PE relationship with eMB

```python
model <- lm(
  eTMB ~ 
    PE_binary + GA +
    Path_MVM_binary + 
    Fetal_sex_binary +
    `age_delivery.x`,
  data = df
)

summary(model)

emm <- emmeans(
  model,
  ~ PE_binary 
)

pairs(emm)
```

```python
# check for functional form specification
resettest(model, power=2:3, type="regressor")

# check for heteroscedasticity
bptest(model)

# check for autocorrelation
dwtest(model)

# check for multicollinearity 
vif(model)
```

```python
model <- lm(
  eTMB ~ 
    PE_binary * GA_binary +
    Path_MVM_binary + 
    Fetal_sex_binary +
    `age_delivery.x`,
  data = df
)

summary(model)

emm <- emmeans(
  model,
  ~ PE_binary * GA_binary
)

pairs(emm)
```
